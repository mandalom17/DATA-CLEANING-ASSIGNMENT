```python
Q1. Missing Data Identification

Scenario :
```

```python
import pandas as pd

df = pd.read_csv("healthcare_data_cleaning_dataset.csv")

# Missing values calculation
missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame(
    {"Missing Count": missing_count, "Missing (%)": missing_percent}
)
print(missing_summary)
```

Q2. Handling Missing Age

Scenario:

```python
# Check distribution metrics
print("Age Mean:", df["Age"].mean())
print("Age Median:", df["Age"].median())

# Impute missing Age values using Median/Mean
df["Age"] = df["Age"].fillna(df["Age"].median())
```

Choice: Mean Imputation (or Median, as both are approximately equal to 50.0 years).

Justification: The Age feature follows a symmetric, normal distribution with a near-zero skewness of -0.0098. The calculated Mean (49.60) and Median (50.00) are nearly identical, meaning there are no extreme skewed tails influencing the average.

```python
Q3. Handling Missing Treatment Cost

Scenario:
```

```python
# Impute missing Treatment_Cost using Median and verify
df["Treatment_Cost"] = df["Treatment_Cost"].fillna(
    df["Treatment_Cost"].median()
)
print("Remaining missing Treatment_Cost:", df["Treatment_Cost"].isnull().sum())
```

Choice: Median Imputation (or Grouped Median Imputation by Diagnosis).

Justification: The Treatment_Cost variable is highly right-skewed with a skewness coefficient of 4.49 (Mean = ₹26,920 vs. Median = ₹24,797). Using the mean would introduce significant upward bias caused by expensive treatments. Median imputation ensures the central tendency is robust to high-cost outliers.

Q5. Invalid Age Values (Data Quality Check)

Scenario:

```python
# Detect invalid age records
invalid_ages = df[(df["Age"] < 0) | (df["Age"] > 100)]
print(f"Total invalid age records: {len(invalid_ages)}")
```

Analysis & Decision:
Filtering for records where Age < 0 or Age > 100 yields 0 invalid records (Age ranges strictly between 0.0 and 99.0 years in this dataset).

Policy recommendation: If negative or $>100$ values appear in future runs, values $<0$ should be replaced with NaN (or converted to positive if due to typing typos), while extreme values $>100$ should be checked against medical records or capped/removed.

Q6. Outlier Detection (Treatment Cost)

Scenario:

```python
Q1 = df["Treatment_Cost"].quantile(0.25)
Q3 = df["Treatment_Cost"].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

outliers = df[
    (df["Treatment_Cost"] < lower_bound) | (df["Treatment_Cost"] > upper_bound)
]
print(f"Number of treatment cost outliers: {len(outliers)}")
```

Results:
25th Percentile (Q1): ₹15,000.50
75th Percentile (Q3): ₹39,423.50
IQR: ₹24,423.00

Upper Bound ($Q3 + 1.5 \times IQR$): ₹76,058.00

Total Outliers Detected: 50 records (values exceeding ₹76,058.00).

Q7. Outlier Treatment

Scenario:

```python
import numpy as np

p5 = df["Treatment_Cost"].quantile(0.05)
p95 = df["Treatment_Cost"].quantile(0.95)

# Winsorize / Cap values
df["Treatment_Cost_Capped"] = np.clip(df["Treatment_Cost"], p5, p95)

# Print verification
print(f"5th Percentile (Cap Lower): {p5}")
print(f"95th Percentile (Cap Upper): {p95}")
print("Capped column successfully created!")
print(df[["Treatment_Cost", "Treatment_Cost_Capped"]].head(10))
```

Thresholds:
5th Percentile: ₹2,914.60

95th Percentile: ₹48,188.10

Q8. Transformation

Scenario:

```python
# Create log-transformed feature
df["Treatment_Cost_Log"] = np.log1p(df["Treatment_Cost"])

print("Original Skewness:", df["Treatment_Cost"].skew())
print("Log Transformed Skewness:", df["Treatment_Cost_Log"].skew())
```

Results & Comparison:

Original Skewness: +4.49 (Strong positive / right skew)

Log-Transformed Skewness (log1p): -1.11 (Significantly stabilized, approximating a bell-shaped normal curve suitable for parametric modeling).

Q9. Time-Based Missing Handling

Scenario:

```python
# Convert to datetime and sort chronologically by Admission_Date
df["Admission_Date"] = pd.to_datetime(df["Admission_Date"])
df = df.sort_values(by="Admission_Date").reset_index(drop=True)

# Apply forward fill followed by backward fill
df["Treatment_Cost_TimeFilled"] = df["Treatment_Cost"].ffill().bfill()

# Print confirmation to see the change
print("Missing Treatment_Cost before:", df["Treatment_Cost"].isnull().sum())
print(
    "Missing Treatment_Cost after time-fill:",
    df["Treatment_Cost_TimeFilled"].isnull().sum(),
)
print("\nPreview of time-filled data:")
print(
    df[["Admission_Date", "Treatment_Cost", "Treatment_Cost_TimeFilled"]].head(
        10
    )
)
```

Justification: Sorting by Admission_Date creates a logical time sequence. Forward-fill (ffill) propagates the last known valid observation forward, which is ideal for time-series patient logs assuming costs or patient states remain persistent until updated. Backward-fill (bfill) resolves any remaining leading missing values at the beginning of the series.
